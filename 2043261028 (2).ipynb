```python
#nomor 1
#output menyatakan total tarif masing masing tipe

bulan=input("Bulan apa Anda akan menginap (Mei, Juni, Juli, Agustus, September, Oktober): ")
lama=int(input("Berapa hari Anda akan menginap: "))
tipe=input("Tipe kamar apa yang Anda pilih (studio atau apartment): ").lower()

tarif_studio=0
tarif_apartment=0
if bulan=="Mei" or bulan=="Oktober":
    tarif_studio=500000
    tarif_apartment=650000
elif bulan=="Juni" or bulan=="September":
    tarif_studio=752000
    tarif_apartment=687000
elif bulan=="Juli" or bulan=="Agustus":
    tarif_studio=760000
    tarif_apartment=770000
else:
    print("Bulan yang Anda masukkan tidak valid.")

total_tarif_studio=0
total_tarif_apartment=0
if tipe=="studio":
    total_tarif_studio=tarif_studio*lama
    if lama>3:
        diskon=total_tarif_studio*0.15
        total_tarif_studio -= diskon
elif tipe=="apartment":
    total_tarif_apartment=tarif_apartment*lama
    if lama>3:
        diskon=total_tarif_apartment*0.10
        total_tarif_apartment -= diskon
else:
    print("Tipe kamar yang Anda masukkan tidak valid.")

print(f'Studio: {total_tarif_studio}')
print(f'Apartment: {total_tarif_apartment}')
```
