```python
pendaftar=[("Andi", 3.45, 85, 3), 
           ("Bela", 3.80, 78, 5), 
           ("Citra", 3.30, 90, 4), 
           ("Dimas", 3.60, 88, 5), 
           ("Eka", 3.10, 92, 3)]
semester_buka=(3, 5)
sudah_asisten=["Dimas"]
semester=True
lolos_nama=[]
lolos_nilai=[]
lolos_semester_5=False
#lolos smt 5 kondisi False karena yang masuk tidak ada yg smt 5
for nama, IPK, nilai, semester in pendaftar:
    syarat_akademik=IPK>=3.25 and nilai>=80 or IPK>=3.0 and nilai>=90
    syarat_semester=semester in semester_buka
    asisten= nama not in sudah_asisten
    print(nama,syarat_akademik,syarat_semester,asisten)
    if syarat_akademik and syarat_semester and asisten:
        lolos_nama.append(nama)
        lolos_nilai.append(nilai)
        if semester == 5:
            lolos_semester_5 = True
print(f'Jumlah lolos  : {len(lolos_nama)}')
print(f'Nama lolos    : {lolos_nama}')
print(f'rata-rata nilai lolos : {sum(lolos_nilai)/len(lolos_nilai)}')
print(f'ada lolos semester 5 : {lolos_semester_5}')

#index output ke 3 yg di print adalah lolos, bukan syarat assistennya, pembenaran di bawah
#di pembenaran bawah Dimas di index akhir adalah Falsa karena syarat ke3 yaitu asisten false (syarat ke3 tdk diprint)
```

```python
pendaftar=[("Andi", 3.45, 85, 3), 
           ("Bela", 3.80, 78, 5), 
           ("Citra", 3.30, 90, 4), 
           ("Dimas", 3.60, 88, 5), 
           ("Eka", 3.10, 92, 3)]
semester_buka=(3, 5)
sudah_asisten=["Dimas"]
semester=True
lolos_nama=[]
lolos_nilai=[]
lolos_semester_5=False
#lolos smt 5 kondisi False karena yang masuk tidak ada yg smt 5
for nama, IPK, nilai, semester in pendaftar:
    syarat_akademik=IPK>=3.25 and nilai>=80 or IPK>=3.0 and nilai>=90
    syarat_semester=semester in semester_buka
    asisten= nama not in sudah_asisten
    lolos=syarat_akademik and syarat_semester and asisten
    print(nama,syarat_akademik,syarat_semester,lolos)
    if syarat_akademik and syarat_semester and asisten:
        lolos_nama.append(nama)
        lolos_nilai.append(nilai)
        if semester == 5:
            lolos_semester_5 = True
print(f'Jumlah lolos          : {len(lolos_nama)}')
print(f'Nama lolos            : {lolos_nama}')
print(f'rata-rata nilai lolos : {sum(lolos_nilai)/len(lolos_nilai)}')
print(f'ada lolos semester 5  : {lolos_semester_5}')
```
