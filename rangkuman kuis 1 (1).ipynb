```python
#INDEXING & SLICING

#indexing=akses karakter tertentu, mulai dari 0
a='statistika bisnis'
print(a[0])

#slicing=mengambil sebagian dari string, mulai nol, kahir -1

print(a[0:3])
#mulai -3 akhir -1-1=-2, peletakan angka tetap sesuai urutan (-3 dulu)
print(a[-3:-1])
#mulai karakter tertentu hingga akhir
print(a[6:])
#mulai awal hingga karakter tertentu
print(a[:6])
#ambil karakter dengan langkah
print(a[::2])
print(a[::3])
```

```python
#Fungsi String (Built-in String Functions)
#pakai print jika dalam satu cell agar terbaca

#upper seluruh karakter kapital
teks='algoritma pemrograman'
print(teks.upper())

#lower seluruh karakter huruf kecil
teks='FISIKA TERAPAN'
print(teks.lower())

#capitalize huruf awal kalimat kapital
teks='algoritma pemrograman'
print(teks.capitalize())

#title huruf awal setiap kata kapital
teks='statistika bisnis'
print(teks.title())

#strip menghilangkan spasi
teks=' hallo '
print(teks.strip())

#replace mengganti substring
teks='hello world'
print(teks.replace('world', 'python'))

#split memisah string dengan pemisah tertentu
teks='hello world'
print(teks.split(' '))

#join menggabnung elemen menjadi satu string
daftar=['hello', 'world']
print(' '.join(daftar))
print('-'.join(daftar))

#find mengembalikan index substring pertama yg ditemukan
#jk tidak ada -1
teks='hello world'
print(teks.find(' world'))
print(teks.find('world'))
print(teks.find('win'))

#count menghitung jumlah kemunculan substring
teks='hello world'
print(teks.count('l'))
print(teks.count('world'))
print(teks.count('win'))
```

```python
a=float(input('Masukkan nilai ke-1: '))
b=float(input('Masukkan nilai ke-2: '))
c=float(input('Masukkan nilai ke-3: '))
d=float(input('Masukkan nilai ke-4: '))
e=float(input('Masukkan nilai ke-5: '))
f=float(input('Masukkan nilai ke-6: '))
data=[a,b,c,d,e,f]

mean=round(sum(data)/len(data), 2)
median=round(sorted(data)[len(data)//2], 2)
range=round(max(data)-min(data), 2)
print(mean, median, range)

dataMax=max(data)
data.remove(dataMax)
mean=round(sum(data)/len(data), 2)
var=round(sum((x-mean)**2 for x in data) / len(data), 2)
std=round(var**0.5, 2)
print(mean, median,var,std)
```
